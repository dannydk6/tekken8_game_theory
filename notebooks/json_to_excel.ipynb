```python
import json
import os
import pandas as pd

def full_path(gbt_filename, base_path=None):
    return gbt_filename if base_path is None else f"{base_path}/{gbt_filename}"

def json_to_excel(filename, base_path=None, output_dir=None):
    scenario_file = full_path(filename, base_path)
    
    # Set path for .json converted file based on output_dir
    out_file = scenario_file.replace('.json', '.xlsx')
    #out_path = f'{output_dir}/{out_file}'
    
    with open(scenario_file, 'r') as f:
        js = json.load(f)
        
    print(js)
    
    sheet1 = [{'key': 'title', 'value': js['title']},
              {'key': 'P1', 'value': js['P1']},
              {'key': 'P2', 'value': js['P2']},
              {'key': 'comment', 'value': js['comment']}]
    sheet1 = pd.DataFrame(sheet1)
    
    print(sheet1)
    
    sheet2 = js['payoffs']
    sheet2 = pd.DataFrame(js['payoffs'])
    sheet2.columns = js['P2-strats']
    sheet2.index  =js['P1-strats']
    sheet2 = sheet2.reset_index()
    sheet2.rename(columns={'index': 'strats'}, inplace=True)
    with pd.ExcelWriter(out_file, engine='xlsxwriter') as writer:
        sheet1.to_excel(writer, sheet_name='Metadata', index=False)
        sheet2.to_excel(writer, sheet_name='Payoffs', index=False)
    return sheet2
    
```
